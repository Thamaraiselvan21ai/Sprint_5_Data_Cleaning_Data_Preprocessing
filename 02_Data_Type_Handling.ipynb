### 1. What are Data Types?
A data type tells Python what kind of information is stored in a variable or DataFrame column.

For example:

age              → numerical

gender           → categorical

churn            → categorical/binary

signup_date      → date/time

customer_id      → string

Real-world example:

In a telecom company:

| Column          | Example    | Appropriate type   |
| --------------- | ---------- | ------------------ |
| age             | 35         | Integer            |
| monthly_charges | 59.99      | Float              |
| gender          | Male       | Categorical        |
| churn           | Yes        | Categorical/Binary |
| signup_date     | 2024-05-20 | DateTime           |
| customer_id     | CUST100001 | String             |



#### AI/ML usage

Correct data types help ML algorithms understand how features should be processed.

For example:

* Numerical → scaling/transformation
* Categorical → encoding
* Date → feature extraction
* Boolean → binary representation
* String → usually cleaned or converted

### 2. Why Are Correct Data Types Important?

This is one of the most important concepts in this notebook.

Suppose we have:

monthly_charges,

"50.5" 
"60.2" 
"75.8"

If Python treats these as strings, mathematical operations may fail or behave incorrectly.

For example:

"50.5" + "60.2" produces: '50.560.2' instead of 110.7

But if they are numeric:

50.5 + 60.2  we get: 110.7

In ML:

Correct data types are important because:

* Mathematical operations require numerical types.
* Scaling requires numerical data.
* Encoding requires categorical data.
* Date analysis requires DateTime.
* Invalid types can cause model errors.
* Incorrect types can produce incorrect analysis.
* Correct types make preprocessing pipelines reliable.

### 12. String → Numeric

This is particularly important for our dataset.

Because we intentionally included: monthly_charges with values such as:

59.99,
70.50,
unknown

we cannot simply use:

astype(float) because "unknown" cannot be converted to a number.

```python
df["monthly_charges"] = pd.to_numeric(
    df["monthly_charges"],
    errors="coerce"
)
```

Now:

"59.99" → 59.99

"70.50" → 70.50

"unknown" → NaN

Why NaN?

Because errors="coerce" tells Pandas:

If conversion fails, replace the invalid value with NaN.

Then we can handle the missing values later.

```python
#for total_charges

df["total_charges"] = pd.to_numeric(
    df["total_charges"],
    errors="coerce"
)
```

```python
df["total_charges"].dtype
```

```python
df.dtypes
```

### 13. to_numeric()

### 14. to_datetime()

pd.to_datetime() converts strings or other date-like values into Pandas DateTime objects.

If an invalid date occurs:

2025-13-50

then: errors="coerce" converts it to: NaT

NaT means Not a Time.

```python
df["signup_date"] = pd.to_datetime(
    df["signup_date"],
    errors="coerce"
)
```

```python
df.dtypes
```

### 19. Complete Data Type Correction for Our Dataset

```python
# Load data

df1 = pd.read_csv("telecom_customer_churn.csv")

# String columns

df1["customer_id"] = df1["customer_id"].astype("string")

# Numeric columns

df1["age"] = pd.to_numeric(df1["age"], errors="coerce")

df1["tenure_months"] = pd.to_numeric(
    df1["tenure_months"],
    errors="coerce"
)

df1["monthly_charges"] = pd.to_numeric(
    df1["monthly_charges"],
    errors="coerce"
)

df1["total_charges"] = pd.to_numeric(
    df1["total_charges"],
    errors="coerce"
)

df1["support_calls"] = pd.to_numeric(
    df1["support_calls"],
    errors="coerce"
)

df1["satisfaction_score"] = pd.to_numeric(
    df1["satisfaction_score"],
    errors="coerce"
)

df1["data_usage_gb"] = pd.to_numeric(
    df1["data_usage_gb"],
    errors="coerce"
)

# Date

df1["signup_date"] = pd.to_datetime(
    df1["signup_date"],
    errors="coerce"
)

# Categorical columns

categorical_columns = [
    "gender",
    "contract_type",
    "internet_service",
    "payment_method",
    "senior_citizen",
    "paperless_billing",
    "region",
    "churn"
]

for col in categorical_columns:
    df1[col] = df1[col].astype("category")
```
