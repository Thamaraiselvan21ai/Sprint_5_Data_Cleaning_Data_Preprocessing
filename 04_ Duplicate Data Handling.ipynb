### Duplicate Data Handling in Pandas

### Exact Duplicates

```python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 101],
    "Name": ["Arun", "Bala", "Arun"],
    "Age": [25, 30, 25],
    "City": ["Chennai", "Madurai", "Chennai"]
}

df = pd.DataFrame(data)

# Identify exact duplicates
df.duplicated()

#True indicates a duplicate row.
```

### Partial Duplicates

```python
df.duplicated(
    subset=["Customer_ID", "Name", "City"]
)
```

### Identifying Duplicates
Identifying duplicates means finding repeated records in a dataset before deciding what to do with them.

Pandas provides the duplicated() function for this purpose.

```python
df.duplicated()
```

```python
#Count duplicate rows
df.duplicated().sum()
```

```python
#Display duplicate rows
df[df.duplicated()]
```

```python
#Display all duplicate occurrences
df[df.duplicated(keep=False)]
```

```python
#Check duplicates based on one column
df.duplicated(subset=["Customer_ID"])
```

### duplicated()
duplicated() is a Pandas function used to identify duplicate rows.

It returns Boolean values (True or False).

```python
df.duplicated(
    subset=None,
    keep="first"
)
```

Understanding keep parameter:

Example dataset:
Customer_ID	Name
101	Arun
102	Bala
101	Arun
101	Arun

Method	            Result
keep='first'	    False, False, True, True
keep='last'	        True, False, True, False
keep=False	        True, False, True, True

Remember:
- first → Keep the first occurrence.
- last → Keep the last occurrence.
- False → Mark all repeated occurrences.

### drop_duplicates()
drop_duplicates() is a Pandas function used to remove duplicate rows from a DataFrame.


```python
df.drop_duplicates(
    subset=None,
    keep="first",
    inplace=False
)
```

```python
#Remove exact duplicates
cleaned_df = df.drop_duplicates()

#Keep the first occurrence
cleaned_df = df.drop_duplicates(
    keep="first")

#Keep the last occurrence
cleaned_df = df.drop_duplicates(
    keep="last")

#Remove all duplicate occurrences
cleaned_df = df.drop_duplicates(
    keep=False)

#Remove duplicates using a selected column
cleaned_df = df.drop_duplicates(
    subset=["Customer_ID"])

###This keeps only one record for each Customer_ID.

#Remove duplicates using multiple columns
cleaned_df = df.drop_duplicates(
    subset=["Name", "Age", "City"])

#Modify the original DataFrame
df.drop_duplicates(
    inplace=True)

```

### Duplicate Detection Based on Selected Columns
Sometimes we don't need to compare every column to identify duplicates.

Instead, we can check duplicates using specific columns.

```python
#Check using Name and City
df.duplicated(
    subset=["Name", "City"])
```

```python
#Remove duplicate customers
cleaned_df = df.drop_duplicates(
    subset=["Customer_ID"],
    keep="first")
cleaned_df
```

### Handling Duplicate Records
Handling duplicate records means identifying, investigating, and deciding how to treat repeated records.

```python
#Common steps
#Step 1: Check total rows
df.shape

#Step 2: Identify duplicate rows
df.duplicated().sum()

#Step 3: Display duplicate records
df[df.duplicated(keep=False)]

#Step 4: Understand duplicates
df[df.duplicated(    subset=["Customer_ID"],    keep=False)]

#Step 5: Remove confirmed duplicates
cleaned_df = df.drop_duplicates()

#Step 6: Verify the result
cleaned_df.duplicated().sum()
```

### Business Rules for Duplicate Removal
Business rules are conditions defined by a company to decide whether a record should be considered a duplicate or a valid record.

We should not remove records only because some values are repeated.
**Business rule:** Order_ID should be unique.

The repeated O101 record can be investigated and a confirmed duplicate removed.
