#  Data Validation
Data Validation is the process of checking whether data follows predefined rules, conditions, and requirements.

```python
import pandas as pd
import numpy as np
data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya', 'Ravi'],
    'Age': [25, 28, -5, 35, 150],
    'Salary': [30000, 35000, -1000, 45000, 50000],
    'Percentage': [85, 92, 105, 75, -10],
    'Gender': ['Male', 'Female', 'Unknown', 'Female', 'Male'],
    'Joining_Date': [
        '2022-01-10',
        '2023-03-15',
        '2021-06-20',
        '2020-05-12',
        '2024-01-10'
    ],
    'Date_of_Birth': [
        '1998-05-10',
        '1996-08-15',
        '2000-02-20',
        '1990-04-12',
        '2010-01-10'
    ]
}

df = pd.DataFrame(data)

print(df)
```

## Range Validation
Range validation checks whether numerical values fall within an acceptable minimum and maximum range.

### Examples:
- Age must be between 18 and 60 for this example.
- Salary cannot be negative.
- Percentage must be between 0 and 100.

```python
# Age validation
invalid_age = df[
    (df['Age'] < 18) | (df['Age'] > 60)
]

print("Invalid Age Records:")
print(invalid_age[['Employee_ID', 'Name', 'Age']])
```

#### Salary validation

```python
invalid_salary = df[df['Salary'] < 0]

print("Invalid Salary Records:")
print(invalid_salary[['Employee_ID', 'Name', 'Salary']])
```

## . Type Validation
Type validation checks whether each column contains the expected data type.
#### Examples:
- Age should be an integer or numerical value.
- Salary should be numerical.
- Joining_Date should be a date.
- Name should be text.
#### Check data types

```python
print(df.dtypes)
```

#### Validate Age type

```python
if pd.api.types.is_numeric_dtype(df['Age']):
    print("Age column type is valid")
else:
    print("Age column type is invalid")
```

#### Convert invalid numerical values

```python
data = pd.DataFrame({
    'Name': ['Arun', 'Priya', 'Kumar'],
    'Age': ['25', '28', 'Twenty']
})

data['Age'] = pd.to_numeric(
    data['Age'],
    errors='coerce'
)

print(data)
```

## Format Validation
Format validation checks whether values follow a required pattern.

### Examples:
- Email should contain a valid email structure.
- Phone number should contain 10 digits (for this example).
- Employee ID should follow a specific format.
#### Email validation

```python
emails = pd.Series([
    'arun@gmail.com',
    'priya@yahoo.com',
    'invalid-email',
    'kumar@gmail.com'
])

email_pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

valid_emails = emails.str.match(email_pattern, na=False)

print(valid_emails)
```

#### Phone number validation

```python
phone = pd.Series([
    '9876543210',
    '12345',
    '9876543210',
    'abcdefghij'
])

valid_phone = phone.str.match(r'^[6-9]\d{9}$', na=False)

print(valid_phone)
```

## Category Validation
Category validation checks whether values belong to an approved list of categories.

### Example:
Gender must be one of the accepted values:
- Male
- Female
- Other

```python
allowed_departments = [
    'IT',
    'HR',
    'Finance',
    'Marketing'
]

department = pd.Series([
    'IT',
    'HR',
    'Sales',
    'Finance'
])

invalid_department = department[
    ~department.isin(allowed_departments)
]

print(invalid_department)
```

## Null Validation
Null validation checks whether required fields contain missing values.

### Examples:
- Employee_ID should not be missing.
- Employee Name should not be missing.
- Salary should not be missing.
#### Identify missing values

```python
print(df.isnull().sum())
```

```python
missing_percentage = (
    df.isnull().mean() * 100
)

print(missing_percentage)
```

```python
missing_names = df[df['Name'].isnull()]

print(missing_names)
```

## Business Rule Validation
Business rule validation checks whether data follows real-world business conditions.

Unlike simple range validation, business rules can compare two or more columns.

### Example rules:
- Salary must be greater than zero.
- Joining date must be after date of birth.
- Employee age must be at least 18.
- Discount should not exceed 100%.
#### Joining Date vs Date of Birth

```python
invalid_dates = df[
    df['Joining_Date'] <= df['Date_of_Birth']
]

print("Invalid Joining Date Records:")
print(invalid_dates[
    ['Employee_ID', 'Name', 'Date_of_Birth', 'Joining_Date']
])
```

#### Salary business rule:

```python
invalid_salary = df[df['Salary'] <= 0]

print(invalid_salary[['Name', 'Salary']])
```

#### Multiple Business Rule:

```python
invalid_records = df[
    (df['Age'] < 18) |
    (df['Salary'] <= 0) |
    (df['Percentage'] < 0) |
    (df['Percentage'] > 100) |
    (df['Joining_Date'] <= df['Date_of_Birth'])
]

print("Records violating business rules:")
print(invalid_records)
```

## Referential Validation
Referential validation checks whether a value in one dataset exists in another related dataset.

### Example,
Every employee's Department_ID should exist in the Department table.

```python
employees = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya'],
    'Department_ID': [1, 2, 3, 5]
})

departments = pd.DataFrame({
    'Department_ID': [1, 2, 3, 4],
    'Department_Name': [
        'IT',
        'HR',
        'Finance',
        'Marketing'
    ]
})
```

## Date Validation
Date validation checks whether date values are valid and follow the required date rules.

### Examples:
- Joining date should not be before date of birth.
- Order date should not be in the future.
- Expiry date should be after manufacturing date.

```python
dates = pd.Series([
    '2024-01-10',
    '2024-02-15',
    'invalid-date'
])

dates = pd.to_datetime(
    dates,
    errors='coerce'
)

print(dates)
```

## Constraint Validation
Constraint validation checks whether data follows strict conditions or restrictions.

### Examples:
- Employee_ID must be unique.
- Employee_ID cannot be null.
- Salary must be positive.
- Age must be within an accepted range.

```python
duplicate_ids = df[
    df['Employee_ID'].duplicated(keep=False)
]

print("Duplicate Employee IDs:")
print(duplicate_ids)
```
