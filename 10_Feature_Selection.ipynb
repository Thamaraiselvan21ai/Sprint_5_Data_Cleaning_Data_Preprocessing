# Notebook 10 – Feature Selection

### What is Feature Selection?

Feature Selection is a data preprocessing technique used to select the most important and relevant features (columns) from a dataset while removing irrelevant, redundant, or less useful features.

It helps improve machine learning model performance by reducing unnecessary information.

##### Mathematical Concept

Suppose a dataset contains `n` features:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

Feature Selection identifies a subset of relevant features:

$$
X_{\text{selected}} \subseteq X
$$

##### Example

Suppose a dataset contains five features:

$$
X = \{X_1, X_2, X_3, X_4, X_5\}
$$

After Feature Selection, only three relevant features are selected:

$$
X_{\text{selected}} = \{X_1, X_2, X_4\}
$$

Therefore:

$$
X_{\text{selected}} \subseteq X
$$

Where:

- `X` = Set of all original features.
- `X_selected` = Subset of selected relevant features.
- `X₁, X₂, ..., Xₙ` = Individual features.
- `n` = Total number of features.

Why Feature Selection Is Important

1. Reduces Overfitting: Removes irrelevant features that may introduce noise.
2. Improves Model Performance: Helps models focus on useful information.
3. Reduces Training Time: Fewer features can reduce computational requirements.
4. Improves Interpretability: Makes it easier to understand model decisions.
5. Removes Redundant Features: Avoids keeping unnecessary duplicate information.
6. Reduces Complexity: Creates simpler machine learning models

Types of Feature Selection:

There are three major approaches:

1. Filter Method

Selects features using statistical techniques without training a machine learning model.

Examples:

- Correlation
- Chi-Square Test
- ANOVA
- Variance Threshold

2. Wrapper Method

Selects features by repeatedly training and evaluating a machine learning model.

Examples:

- Forward Selection
- Backward Elimination
- Recursive Feature Elimination (RFE)

3. Embedded Method

Performs feature selection during the model training process.

Examples:

- Lasso (L1 Regularization)
- Decision Tree Feature Importance
- Random Forest Feature Importance

Real-World Applications:

- Banking: Select important customer features to predict loan defaults.
- Healthcare: Select important medical features to predict diseases.
- E-commerce: Identify important customer features to predict purchases.
- Fraud Detection: Select transaction features to identify fraudulent transactions

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_classif

# Sample dataset
data = {
    "Age": [25, 30, 35, 28, 40, 45, 22, 38],
    "Salary": [25000, 40000, 55000, 30000,
               70000, 80000, 20000, 65000],
    "Experience": [2, 5, 8, 3, 12, 15, 1, 10],
    "Spending_Score": [80, 30, 90, 40, 85, 25, 70, 35],
    "Purchased": [1, 0, 1, 0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Purchased", axis=1)
y = df["Purchased"]

# Select top 3 features
selector = SelectKBest(
    score_func=f_classif,
    k=3
)

X_selected = selector.fit_transform(X, y)

# Display selected features
selected_features = X.columns[selector.get_support()]

print("Selected Features:")
print(selected_features.tolist())

print("\nSelected Dataset:")
print(pd.DataFrame(
    X_selected,
    columns=selected_features
))
```

Code explanation:

1. SelectKBest: Selects the top K features.
2. f_classif: Uses the ANOVA F-test to calculate feature relevance for classification.
3. k=3: Selects the three highest-scoring features.
4. fit_transform(): Calculates scores and transforms the dataset.
5. get_support(): Identifies the selected features.

### Relevant Features


Relevant Features are the features (independent variables) in a dataset that contain useful information for predicting or explaining the target variable.

These features help a Machine Learning model identify patterns and make accurate predictions.

In simple words, relevant features are the columns that provide useful information for solving a particular problem


##### Mathematical Concept

Suppose a dataset contains \\(n\\) features:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

The target variable is:

$$
Y = \text{Target Variable}
$$

Relevant features are the features that provide useful information about the target variable.

$$
X_{\text{relevant}} \subseteq X
$$

Where:

- \(X\) = All available features.
- \(X_{\text{relevant}}\) = Selected relevant features.
- \(Y\) = Target variable.
- \(n\) = Total number of features.


```python

import pandas as pd
from sklearn.feature_selection import SelectKBest, f_classif

# Step 1: Create Dataset
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9],
    'Attendance': [50, 55, 60, 65, 70, 80, 90, 95],
    'Previous_Marks': [40, 45, 50, 55, 60, 70, 80, 90],
    'Assignment_Score': [35, 40, 50, 55, 65, 75, 85, 95],
    'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Random_Number': [91, 12, 75, 33, 81, 22, 64, 10],
    'Exam_Result': [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Separate Features and Target
X = df.drop('Exam_Result', axis=1)
y = df['Exam_Result']

# Step 3: Select Relevant Features
selector = SelectKBest(score_func=f_classif, k=4)

X_selected = selector.fit_transform(X, y)

# Step 4: Get Selected Feature Names
selected_features = X.columns[selector.get_support()]

print("\nRelevant Features:")
print(selected_features.tolist())

# Step 5: Display Feature Scores
feature_scores = pd.DataFrame({
    'Feature': X.columns,
    'Score': selector.scores_
})

feature_scores = feature_scores.sort_values(
    by='Score', ascending=False
)

print("\nFeature Scores:")
print(feature_scores)

# Step 6: Display Selected Dataset
selected_df = pd.DataFrame(
    X_selected,
    columns=selected_features
)

print("\nDataset After Feature Selection:")
print(selected_df)
```

Understanding the Program

- X – Contains independent features.
- y – Contains the target variable.
- SelectKBest() – Selects the top 4 features.
- f_classif – Uses the ANOVA F-test to calculate feature scores for classification.
- get_support() – Identifies selected features.
- scores_ – Returns the calculated feature scores.

### Irrelevant Features


Irrelevant Features are the features (columns) in a dataset that do not provide useful information for predicting or explaining the target variable.

These features do not contribute meaningful information to a Machine Learning model and may introduce unnecessary noise.

In simple words, irrelevant features are columns that are not useful for solving a particular prediction problem.


##### Mathematical Concept

Suppose a dataset contains \\(n\\) features:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

The target variable is:

$$
Y = \text{Target Variable}
$$

Irrelevant features are the features that provide little or no useful information about the target variable.

$$
X_{\text{irrelevant}} \subseteq X
$$

Where:

- \(X\) = All available features.
- \(X_{\text{irrelevant}}\) = Irrelevant features.
- \(Y\) = Target variable.
- \(n\) = Total number of features.


```python

import pandas as pd
from sklearn.feature_selection import VarianceThreshold

# Step 1: Create Dataset
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9],
    'Attendance': [50, 55, 60, 65, 70, 80, 90, 95],
    'Previous_Marks': [40, 45, 50, 55, 60, 70, 80, 90],
    'Assignment_Score': [35, 40, 50, 55, 65, 75, 85, 95],
    'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Random_Number': [91, 12, 75, 33, 81, 22, 64, 10],
    'Constant_Column': [1, 1, 1, 1, 1, 1, 1, 1],
    'Exam_Result': [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Separate Features and Target
X = df.drop('Exam_Result', axis=1)
y = df['Exam_Result']

# Step 3: Identify Constant Features
selector = VarianceThreshold(threshold=0)

X_filtered = selector.fit_transform(X)

# Step 4: Get Remaining Features
remaining_features = X.columns[selector.get_support()]

# Step 5: Identify Constant Features
irrelevant_features = X.columns[~selector.get_support()]

print("\nConstant Features Removed:")
print(irrelevant_features.tolist())

print("\nRemaining Features:")
print(remaining_features.tolist())

# Step 6: Create Filtered Dataset
filtered_df = pd.DataFrame(
    X_filtered,
    columns=remaining_features
)

filtered_df['Exam_Result'] = y.values

print("\nDataset After Filtering:")
print(filtered_df)
```

Understanding the Program

- VarianceThreshold() identifies features with zero or low variance.
- threshold=0 removes features whose values are constant.
- get_support() identifies the features that remain.
- ~selector.get_support() identifies the removed features.

Why Remove Irrelevant Features?

1. Reduce unnecessary information.
2. Reduce model complexity.
3. Improve training efficiency.
4. Help reduce overfitting.
5. Improve model interpretability.
6. Reduce noise in the dataset.

### Redundant Features

Redundant Features are the features (columns) in a dataset that contain duplicate or highly similar information already provided by other features.

These features provide little additional information to a Machine Learning model because their information overlaps with other features.

In simple words, redundant features are columns that repeat almost the same information as other columns.


##### Mathematical Concept

Suppose a dataset contains \\(n\\) features:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

Two features are considered highly correlated when their correlation coefficient is close to +1 or -1.

$$
r(X_i, X_j) \approx +1 \quad \text{or} \quad -1
$$

Where:

- \(X_i\) = First feature.
- \(X_j\) = Second feature.
- \(r\) = Correlation coefficient.
- \(+1\) = Perfect positive correlation.
- \(-1\) = Perfect negative correlation.

Important: Highly correlated features may contain redundant information, but correlation alone does not always mean that one feature should be removed.


```python

import pandas as pd
import numpy as np

# Step 1: Create Dataset
data = {
    'Age': [22, 25, 28, 30, 35, 40, 45, 50],
    'Experience': [1, 3, 5, 7, 10, 12, 15, 18],
    'Monthly_Salary': [20000, 25000, 30000, 35000,
                       40000, 45000, 50000, 55000],
    'Annual_Salary': [240000, 300000, 360000, 420000,
                      480000, 540000, 600000, 660000],
    'Height': [160, 165, 170, 168, 175, 172, 180, 178],
    'Performance_Score': [60, 65, 70, 72, 78, 82, 88, 92]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Calculate Correlation Matrix
correlation_matrix = df.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix.round(2))

# Step 3: Identify Highly Correlated Features
threshold = 0.90

upper_triangle = correlation_matrix.abs().where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

redundant_features = [
    column for column in upper_triangle.columns
    if any(upper_triangle[column] > threshold)
]

print("\nRedundant Features:")
print(redundant_features)

# Step 4: Remove Redundant Features
df_selected = df.drop(columns=redundant_features)

print("\nDataset After Removing Redundant Features:")
print(df_selected)
```

Why Remove Redundant Features?

1. Reduce unnecessary features.
2. Reduce multicollinearity.
3. Make models easier to interpret.
4. Reduce model complexity.
5. Improve computational efficiency.
6. Avoid giving excessive weight to duplicated information in some models.

### Correlation-Based Selection


Correlation-Based Feature Selection is a feature selection technique that uses correlation coefficients to identify relationships between features and the target variable.

It selects features that have a strong relationship with the target variable and removes features that are highly correlated with each other when they provide redundant information.

In simple words, it helps us identify which features are useful for prediction and which features repeat similar information.


##### Mathematical Concept

The Pearson correlation coefficient measures the strength and direction of a linear relationship between two variables.

$$
r(X,Y)=
\frac{\operatorname{Cov}(X,Y)}
{\sigma_X \sigma_Y}
$$

Where:

- \(r(X,Y)\) = Correlation coefficient between X and Y.
- \(\operatorname{Cov}(X,Y)\) = Covariance between X and Y.
- \(\sigma_X\) = Standard deviation of feature X.
- \(\sigma_Y\) = Standard deviation of target Y.


Types of Correlation-Based Selection

1. Feature-to-Target Correlation: Select features that have a strong correlation with the target variable.
2. Feature-to-Feature Correlation: Identify highly correlated features and remove redundant ones.

```python

import pandas as pd
import numpy as np

# Step 1: Create Dataset
np.random.seed(42)

data = {
    'Experience': np.random.randint(1, 20, 100),
    'Education': np.random.randint(1, 6, 100),
    'Age': np.random.randint(22, 55, 100),
    'Employee_ID': range(101, 201),
    'Random_Number': np.random.randint(1, 1000, 100)
}

df = pd.DataFrame(data)

# Create Salary Target
df['Salary'] = (
    15000
    + df['Experience'] * 3000
    + df['Education'] * 2000
    + np.random.normal(0, 3000, 100)
)

print("Original Dataset:")
print(df.head())

# Step 2: Calculate Feature-to-Target Correlation
target_corr = df.corr(numeric_only=True)['Salary'].drop('Salary')

print("\nCorrelation with Salary:")
print(target_corr.sort_values(ascending=False))

# Step 3: Select Relevant Features
threshold = 0.30

relevant_features = target_corr[
    target_corr.abs() >= threshold
].index.tolist()

print("\nRelevant Features:")
print(relevant_features)

# Step 4: Calculate Feature-to-Feature Correlation
X = df[relevant_features]

corr_matrix = X.corr().abs()

# Get Upper Triangle
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Step 5: Identify Redundant Features
redundant_features = [
    column for column in upper_triangle.columns
    if any(upper_triangle[column] > 0.85)
]

print("\nRedundant Features:")
print(redundant_features)

# Step 6: Remove Redundant Features
selected_features = [
    feature for feature in relevant_features
    if feature not in redundant_features
]

print("\nFinal Selected Features:")
print(selected_features)

# Step 7: Final Dataset
final_df = df[selected_features + ['Salary']]

print("\nDataset After Feature Selection:")
print(final_df.head())
```

Advantages:

1. Easy to understand and implement.
2. Reduces the number of features.
3. Helps identify relevant features.
4. Helps detect redundant features.
5. Reduces model complexity.

Limitations:

1. Pearson correlation mainly identifies linear relationships.
2. It may miss nonlinear relationships.
3. Correlation does not imply causation.
4. A feature with low individual correlation may still be useful when combined with other features.
5. Correlation-based selection is not always suitable for categorical features without appropriate encoding.

### Variance Threshold

ariance Threshold is a Filter-Based Feature Selection technique used to remove features that have very low variance or zero variance.

It identifies features whose values do not change much across observations and removes them because they may provide limited useful information to a Machine Learning model.

In simple words, Variance Threshold removes columns that contain constant or almost constant values.


##### Mathematical Concept

Variance measures how much the values of a feature differ from its mean.

The population variance formula is:

$$
\operatorname{Var}(X) = E(X^2) - [E(X)]^2
$$

For a normally distributed variable:

$$
\operatorname{Var}(X) = \sigma^2
$$

Suppose the standard deviation is 1.4:

$$
\operatorname{Var}(X) = (1.4)^2 = 1.96
$$

Where:

- \(X\) = Feature values.
- \(\mu\) = Mean of the feature.
- \(\sigma\) = Standard deviation.
- \(N\) = Total number of observations.
- \(\operatorname{Var}(X)\) = Variance of the feature.
- \(E(X)\) = Expected value (mean).
- \(E(X^2)\) = Expected value of squared values.

##### Variance Threshold Condition

$$
\operatorname{Var}(X) \leq T
$$

Where:

- \(\operatorname{Var}(X)\) = Feature variance.
- \(T\) = Selected variance threshold.

If the variance is less than or equal to the threshold, the feature is removed.

Otherwise, the feature is retained.


```python

import pandas as pd
from sklearn.feature_selection import VarianceThreshold

# Step 1: Create Dataset
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9],
    'Attendance': [50, 60, 70, 80, 90, 85, 75, 95],
    'Previous_Marks': [40, 45, 50, 55, 60, 65, 70, 75],
    'Gender_Code': [1, 1, 1, 1, 1, 1, 1, 1],
    'Constant_Column': [5, 5, 5, 5, 5, 5, 5, 5],
    'Exam_Result': [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Separate Features and Target
X = df.drop('Exam_Result', axis=1)
y = df['Exam_Result']

# Step 3: Calculate Feature Variance
print("\nFeature Variance:")
print(X.var())

# Step 4: Apply Variance Threshold
selector = VarianceThreshold(threshold=0)

X_selected = selector.fit_transform(X)

# Step 5: Identify Selected Features
selected_features = X.columns[selector.get_support()]

removed_features = X.columns[~selector.get_support()]

print("\nSelected Features:")
print(selected_features.tolist())

print("\nRemoved Features:")
print(removed_features.tolist())

# Step 6: Create Final Dataset
final_df = pd.DataFrame(
    X_selected,
    columns=selected_features
)

final_df['Exam_Result'] = y.values

print("\nDataset After Variance Threshold:")
print(final_df)
```

Advantages:

1. Easy to implement.
2. Removes constant features.
3. Reduces dataset dimensions.
4. Reduces unnecessary computation.
5. Helps simplify Machine Learning models
   
Limitations:

1. Does not consider the target variable.
2. Cannot determine whether a feature is important for prediction.
3. Low variance does not always mean a feature is irrelevant.
4. Features with different scales may have different variance values.

### Univariate Feature Selection

Univariate Feature Selection is a feature selection technique that evaluates each feature independently to determine how strongly it is related to the target variable.

It selects the most important features based on a statistical score.

In simple words, it checks one feature at a time to identify which features are useful for predicting the target.

##### Mathematical Concept

Suppose a dataset contains 3 features:

$$
X = \{X_1, X_2, X_3\}
$$

Univariate Feature Selection checks each feature separately and calculates its score.

$$
S_i = f(X_i, Y)
$$

##### Where:

- $X_i$ = Individual feature (Age, Salary, etc.).
- $Y$ = Target variable (Purchase).
- $f$ = Statistical scoring method.
- $S_i$ = Feature score.

The features with the highest scores are selected.

$$
X_{selected} = \text{Top } k \text{ features}
$$

- $k$ = Number of features we want to select.

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_classif

# Create a sample dataset
data = {
    "Age": [22, 25, 35, 40, 45, 28, 38, 50],
    "Salary": [20000, 30000, 50000, 60000, 70000, 25000, 55000, 80000],
    "Website_Visits": [2, 5, 10, 12, 15, 3, 11, 18],
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Purchase": [0, 0, 1, 1, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Purchase", axis=1)
y = df["Purchase"]

# Select the top 2 features
selector = SelectKBest(score_func=f_classif, k=2)

X_selected = selector.fit_transform(X, y)

# Display feature scores
scores = pd.DataFrame({
    "Feature": X.columns,
    "Score": selector.scores_,
    "Selected": selector.get_support()
})

print("Feature Selection Scores:")
print(scores.sort_values(by="Score", ascending=False))

print("\nSelected Features:")
print(X.columns[selector.get_support()].tolist())
```

Understanding the Code:

- SelectKBest selects the top K features based on statistical scores.
- f_classif uses the ANOVA F-test for classification problems.
- k=2 means the two highest-scoring features are selected.
- fit_transform() calculates scores and returns the selected features.
- get_support() identifies which features were selected.
  
Advantages:

1. Easy to understand and implement.
2. Fast for datasets with many features.
3. Helps remove less informative features.
4. Reduces the number of input features
   
Limitations

1. Evaluates each feature independently.
2. May miss relationships between multiple features.
3. Does not automatically detect all redundant features.
4. Results depend on the statistical scoring method.

### Mutual Information


Mutual Information (MI) is a feature selection technique that measures how much information one feature provides about the target variable.

In simple words, it measures how useful a feature is for predicting the target

- Higher Mutual Information → More information about the target.
- Lower Mutual Information → Less information about the target.
- Zero Mutual Information → No statistical dependence detected.

##### Mathematical Concept

Mutual Information measures how much information a feature $X$ provides about the target variable $Y$.

The formula is:

$$
I(X;Y) = \sum_{x,y} P(x,y)
\log\left(\frac{P(x,y)}{P(x)P(y)}\right)
$$

##### Where:

- $I(X;Y)$ = Mutual Information between X and Y.
- $X$ = Input feature.
- $Y$ = Target variable.
- $P(x,y)$ = Joint probability of X and Y.
- $P(x)$ = Probability of X.
- $P(y)$ = Probability of Y.
- $\log$ = Logarithm.

```python
import os
os.environ["LOKY_MAX_CPU_COUNT"] = "4"

import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
```

```python
import pandas as pd
from sklearn.feature_selection import mutual_info_classif

# Create sample dataset
data = {
    "Age": [22, 25, 35, 40, 45, 28, 38, 50],
    "Salary": [20000, 30000, 50000, 60000, 70000, 25000, 55000, 80000],
    "Website_Visits": [2, 3, 10, 12, 15, 4, 11, 18],
    "Purchase": [0, 0, 1, 1, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Purchase", axis=1)
y = df["Purchase"]

# Calculate Mutual Information scores
mi_scores = mutual_info_classif(
    X, y, random_state=42
)

# Display scores
results = pd.DataFrame({
    "Feature": X.columns,
    "Mutual_Information_Score": mi_scores
})

results = results.sort_values(
    by="Mutual_Information_Score",
    ascending=False
)

print(results)
```

Understanding the Output:

The output displays a Mutual Information score for each feature.

For example, if a feature has a higher MI score, it provides more information about the target variable.

Note: Actual scores depend on the dataset and estimation method

Advantages:

1. Can identify linear and non-linear relationships.
2. Measures the information shared between a feature and target.
3. Useful for feature selection.
4. Can be used for classification and regression with the appropriate scoring function
   
Limitations:

1. Does not show whether the relationship is positive or negative.
2. MI scores can be affected by sample size.
3. Requires suitable estimation methods for numerical and categorical data.
4. Evaluates features individually and may miss interactions between features.

### Recursive Feature Elimination


Recursive Feature Elimination (RFE) is a feature selection technique that repeatedly trains a machine learning model and removes the least important features until the required number of features remains.

In simple words:

RFE starts with all features, identifies the least important feature, removes it, and repeats the process until only the required features are left.
2. How RFE Works
1. Start with all available features.
2. Train a machine learning model using all features.
3. Calculate the importance of each feature.
4. Remove the least important feature.
5. Repeat the process until the required number of features remains.

##### Mathematical Concept

Suppose a dataset contains $n$ features:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

A machine learning model calculates the importance of each feature:

$$
I_i = \text{Importance}(X_i)
$$

RFE removes the feature with the lowest importance:

$$
X_{\text{removed}} = \arg\min_{X_i} I_i
$$

The process repeats until the required number of features $k$ remains.

$$
|X_{\text{selected}}| = k
$$

##### Where:

- $X_i$ = Individual feature.
- $I_i$ = Importance score of a feature.
- $\arg\min$ = Identifies the feature with the lowest importance.
- $k$ = Required number of selected features.
- $X_{\text{selected}}$ = Final selected features.
- $X_{\text{removed}}$ = Feature removed during the process.

```python
import pandas as pd
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier

# Create sample dataset
data = {
    "Age": [22, 25, 35, 40, 45, 28, 38, 50, 30, 42],
    "Salary": [20000, 30000, 50000, 60000, 70000,
               25000, 55000, 80000, 35000, 65000],
    "Website_Visits": [2, 3, 10, 12, 15, 4, 11, 18, 5, 14],
    "Customer_ID": [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],
    "Purchase": [0, 0, 1, 1, 1, 0, 1, 1, 0, 1]
}

df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Purchase", axis=1)
y = df["Purchase"]

# Create machine learning model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Apply Recursive Feature Elimination
rfe = RFE(
    estimator=model,
    n_features_to_select=2
)

rfe.fit(X, y)

# Display feature selection results
results = pd.DataFrame({
    "Feature": X.columns,
    "Ranking": rfe.ranking_,
    "Selected": rfe.support_
})

print("RFE Feature Selection Results:")
print(results)

# Display selected features
print("\nSelected Features:")
print(X.columns[rfe.support_].tolist())
```

Understanding the Code:

- RandomForestClassifier → Machine learning model used to estimate feature importance.
- RFE() → Creates the Recursive Feature Elimination selector.
- n_features_to_select=2 → Selects the best two features.
- fit() → Trains the model and performs recursive feature elimination.
- ranking_ → Displays feature rankings. Rank 1 means selected.
- support_ → Shows whether a feature is selected (True) or removed (False).
  
Advantages:

1. Automatically selects important features.
2. Reduces the number of input features.
3. Can improve model simplicity.
4. Works with different models that provide feature importance or coefficients
   
 Limitations:
 
1. Can be computationally expensive for large datasets.
2. Requires repeated model training.
3. Results depend on the selected machine learning model.
4. May remove useful features if the model estimates importance poorly.

### Feature Importance

Feature Importance measures how much each input feature contributes to a machine learning model's predictions.

In simple words, it tells us which features are more useful for making predictions.

For example, when predicting customer purchases:
- Age may have some importance.
- Salary may have high importance.
- Website Visits may have very high importance.
- Customer ID may have very low importance.

##### Mathematical Concept

Suppose a dataset contains $n$ features:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

A machine learning model calculates an importance score for each feature:

$$
I_i = \text{Importance}(X_i)
$$

The features can be ranked based on their importance scores:

$$
I_1 \geq I_2 \geq I_3 \geq \ldots \geq I_n
$$

##### Where:

- $X_i$ = Individual input feature.
- $I_i$ = Importance score of the feature.
- $n$ = Total number of features.

A higher importance score generally indicates a greater contribution to the model's predictions.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier

# Create sample dataset
data = {
    "Age": [22, 25, 35, 40, 45, 28, 38, 50, 30, 42],
    "Salary": [20000, 30000, 50000, 60000, 70000,
               25000, 55000, 80000, 35000, 65000],
    "Website_Visits": [2, 3, 10, 12, 15, 4, 11, 18, 5, 14],
    "Customer_ID": [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],
    "Purchase": [0, 0, 1, 1, 1, 0, 1, 1, 0, 1]
}

df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Purchase", axis=1)
y = df["Purchase"]

# Create Random Forest model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
model.fit(X, y)

# Calculate feature importance
importance = model.feature_importances_

# Create feature importance table
results = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance
})

# Sort by importance
results = results.sort_values(
    by="Importance",
    ascending=False
)

print("Feature Importance:")
print(results)

# Visualize feature importance
plt.figure(figsize=(8, 5))

plt.barh(
    results["Feature"],
    results["Importance"]
)

plt.xlabel("Importance Score")
plt.ylabel("Features")
plt.title("Random Forest Feature Importance")
plt.gca().invert_yaxis()

plt.show()
```
